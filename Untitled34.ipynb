```python
from abc import ABC, abstractmethod

class DocumentElement(ABC):
  """
  Abstraction for document elements
  Every element in the document must know how to render itself.
  """

  @abstractmethod
  def render(self) -> str:
    """Render the element to a string."""
    pass

class Persistence(ABC):
  """
  Abstraction for document persistence.
  Different storage implementations can follow this contract.
  """

  @abstractmethod
  def save(self, data: str) -> None:
    """Save document data to storage. """
    pass

```

```python
class TextElement(DocumentElement):
  """Concrete implementation for text elements."""

  def __init__(self, text: str):
    self._text = text

  def render(self) -> str:
    return self._text

class ImageElement(DocumentElement):
  """Concrete implementation for image elements."""

  def __init__(self, image_path: str):
    self._image_path = image_path

  def render(self) -> str:
    return f"[Image: {self._image_path}]"

class NewLineElement(DocumentElement):
  def render(self) -> str:
    return "\n"
```

```python
from typing import List
class Document:
  """
  Document class responsible for holding a collection of elements.
  Acts as a container and renderer for all document elements.
  """

  def __init__(self):
    self._elements: List[DocumentElement] = []

  def add_element(self, element:DocumentElement) -> None:
    self._elements.append(element)

  def render(self) -> str:
    result = ""
    for element in self._elements:
      result += element.render()

    return result

```

```python
class FileStorage(Persistence):

  def __init__(self, filename: str = "document.txt"):
    self._filename = filename

  def save(self, data: str) -> None:
    try:
      with open(self._filename, "w") as file:
        file.write(data)
        print(f"Document saved to {self._filename}")

    except IOError as e:
      print(f"Error: Unable to save file. {str(e)}")


class DBStorage(Persistence):

  def __init__(self, connection_string: str =""):
    self._connection_String = connection_string

  def save(self, data:str) -> None:
    print(f"Document saved to database.")


```

```python
class DocumentEditor:
  def __init__(self, document: Document, storage: Persistence) -> None:
    self._document = document
    self._storage = storage
    self._rendered_document: str = ""

  def add_text(self, text:str) -> None:
    self._document.add_element(TextElement(text))
    self._rendered_document= ""

  def add_image(self, image_path: str) -> None:
    self._document.add_element(ImageElement(image_path))
    self._rendered_document =""

  def add_new_line(self) -> None:
    """Add a new line to the document."""
    self._document.add_element(NewLineElement())
    self._rendered_document = ""  # Invalidate cache

  def render_document(self) -> str:
    if not self._rendered_document:
      self._rendered_document = self._document.render()

    return self._rendered_document

  def save_document(self) -> None:
    self._storage.save(self.render_document())
```

```python
def main():
  print("=" * 60)
  print("DOCUMENT EDITOR - PYTHON VERSION")
  print("=" * 60)
  print()

  #Dependency Injection: Create document and storage
  document =  Document()
  storage = FileStorage("output_document.txt")

  editor = DocumentEditor(document, storage)

  print("###Building document...###")
  editor.add_text("Hello, world!")
  editor.add_new_line()
  editor.add_text("This is a real world document editor example.")
  editor.add_new_line()
  editor.add_image("picture.jpg")
  editor.add_new_line()
  editor.add_text("this is the last text for this document.")

  print("###Rendered Document: ###")
  print("-" * 60)
  print(editor.render_document())
  print("-"*60)
  print()

  print("### Saving Document... ###")
  editor.save_document()
  print()
```

```python
if __name__ == "__main__":
    # Run main example
    main()
```

```python
def demonstrate_multiple_documents():
    """
    Shows how multiple documents can work independently.
    """
    print("=" * 60)
    print("MULTIPLE DOCUMENTS DEMONSTRATION")
    print("=" * 60)
    print()

    # Create first document
    doc1 = Document()
    editor1 = DocumentEditor(doc1, FileStorage("document1.txt"))
    editor1.add_text("Document 1")
    editor1.add_new_line()
    editor1.add_text("This is the first document.")
    print("Document 1 saved")
    editor1.save_document()

    # Create second document
    doc2 = Document()
    editor2 = DocumentEditor(doc2, FileStorage("document2.txt"))
    editor2.add_text("Document 2")
    editor2.add_new_line()
    editor2.add_text("This is the second document.")
    print("Document 2 saved")
    editor2.save_document()

    print()

demonstrate_multiple_documents()
```
